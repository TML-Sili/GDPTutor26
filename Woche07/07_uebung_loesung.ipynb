## Aufgabe 3 (4 Punkte)
Schreiben Sie eine Methode `nonSquareDivisible` welche zwei ganzzahlige Parameter `n` und `k` entgegennimmt und alle ganzen Zahlen zwischen 1 und $n$ ausgibt, die durch $k$ teilbar sind, aber keine Quadratzahl sind.

Die Quadratwurzelprüfung schliesst jede Quadratzahl aus, nicht nur `k * k`.


````java
class Numbers {
    public static void nonSquareDivisible(int n, int k) {
        for (int i = 1; i <= n; i++) {
            int wurzel = (int) Math.sqrt(i);
            if (i % k == 0 && wurzel * wurzel != i) {
                System.out.println(i);
            }
        }
    }

    public static void main(String[] args) {
        System.out.println("n = 16, k = 3:");
        nonSquareDivisible(16, 3);
        System.out.println("n = 20, k = 2:");
        nonSquareDivisible(20, 2);
    }
}

Numbers.main(new String[0]);
````

## Aufgabe 5 (4 Punkte)
Schreiben Sie eine Methode `drawIdentity` welche eine positive ganze Zahl `n` entgegennimmt und die Identitätsmatrix von Grösse $n$ ausgibt. (die Identitätsmatrix eine Tabelle mit $n$ Zeilen und $n$ Spalten, wobei fast überall eine 0 steht, ausser auf der Diagonale, d.h. da wo die Zeilennummer und die Spaltennummer gleich sind. Dort steht eine 1.) Für $n=4$ sollten Sie die folgende Matrix erhalten:

```
1000
0100
0010
0001
```

Die Diagonale hat gleiche Zeilen- und Spaltenindizes.


````java
class Matrix {
    public static void drawIdentity(int n) {
        for (int zeile = 0; zeile < n; zeile++) {
            for (int spalte = 0; spalte < n; spalte++) {
                if (zeile == spalte) {
                    System.out.print(1);
                } else {
                    System.out.print(0);
                }
            }
            System.out.println();
        }
    }

    public static void main(String[] args) {
        drawIdentity(4);
    }
}

Matrix.main(new String[0]);
````

## Aufgabe 6 (3 Punkte)
Schreiben Sie eine Methode `fibonacci` welche die $n$'te Fibonacci-Zahl $f_n$ zurückgibt. Die Fibonacci-Zahlen sind wie folgt definiert:
* $f_0 = 1$
* $f_1 = 1$
* $f_n = f_{n-1} + f_{n-2}$.

Beide Anfangswerte sind `1`; danach werden die zwei vorherigen Werte addiert.


````java
class Fibonacci {
    public static int fibonacci(int n) {
        if (n == 0 || n == 1) {
            return 1;
        }
        return fibonacci(n - 1) + fibonacci(n - 2);
    }

    public static void main(String[] args) {
        System.out.println(fibonacci(0));
        System.out.println(fibonacci(1));
        System.out.println(fibonacci(6));
    }
}

Fibonacci.main(new String[0]);
````

## Aufgabe 7 (3 Punkte)

Schreiben Sie eine Methode `assignments`, die zwei ganze Zahlen entgegen nimmt: die Anzahl Studierenden und die maximale Grösse einer Übungsgruppe. Die Methode soll berechnen, wieviele Übungsgruppen mindestens erstellt werden müssen, damit alle Studierenden in einer Gruppe unterkommen und keine Gruppe zu gross ist. Die Methode soll die Lösung in einem Satz der Form `Es braucht mindestens X Übungsgruppen` ausgeben, wobei `X` durch den korrekten Wert ersetzt werden soll. Zudem soll `X` als ganze Zahl angezeigt werden, also z.B. als `2`, und nicht als `2.0`.

Die Ganzzahldivision liefert volle Gruppen; bei einem Rest wird eine weitere Gruppe benötigt.


````java
class Groups {
    public static void assignments(int studierende, int maxGroesse) {
        int gruppen = studierende / maxGroesse;
        if (studierende % maxGroesse != 0) {
            gruppen++;
        }
        System.out.println("Es braucht mindestens " + gruppen + " Übungsgruppen");
    }

    public static void main(String[] args) {
        assignments(135, 7);
    }
}

Groups.main(new String[0]);
````

## Aufgabe 8 (3 Punkte)
Schreiben Sie eine Methode `add` welche ein Gleitkommazahl-Arrays und eine Gleitkommazahl `x` entgegennimmt und ein Array zurückgibt, dessen $i$'ter Eintrag die Summe des jeweiligen Eintrags mit `x` ist. Der Aufruf `add({1.2, 2.5, 1.0}, 2.0)` sollte also das Array `{3.2, 4.5, 3.0}` zurückgeben.

Ein neues Array schützt die Eingabewerte.


````java
import java.util.Arrays;

class ArraySum {
    public static double[] add(double[] zahlen, double x) {
        double[] ergebnis = new double[zahlen.length];
        for (int i = 0; i < zahlen.length; i++) {
            ergebnis[i] = zahlen[i] + x;
        }
        return ergebnis;
    }

    public static void main(String[] args) {
        System.out.println(Arrays.toString(add(new double[]{1.2, 2.5, 1.0}, 2.0)));
    }
}

ArraySum.main(new String[0]);
````

## Aufgabe 9 (5 Punkte)
Für diese Aufgabe geben wir Ihnen bereits eine Methode, die Sie weder verstehen noch ändern müssen. Die Methode `cross(x1, y1, x2, y2, x3, y3, x4, y4)` schaut sich die vier Punkte $p_1 = (x_1, y_1)$ ,$p_2 = (x_2, y_2)$, $p_3 = (x_3, y_3)$ und $p_4 = (x_4, y_4)$ in der Ebene an und gibt zurück, ob sich die zwei Segmente $p_1p_2$ und $p_3p_4$ schneiden. Wenn ja, gibt die Methode `true` zurück, ansonsten `false`.

In der Methode `main` sind ausserdem 7 Punkte $p_1,\ldots,p_7$ als Arrays von zwei Gleitkommazahlen definiert. Ihre Aufgabe ist die folgende:
* Schreiben Sie eine Methode `thrackle`, welche ein Array von Punkten (gespeichert als Arrays von zwei Gleitkommazahlen) entgegennimmt und `true` zurückgibt, wenn vier aufeinanderfolgende Punkte $p_1, p_2, p_3, p_4$ im Array immer zwei kreuzende Segmente $p_1p_2$ und $p_3p_4$ bilden. Wenn es vier aufeinanderfolgende Punkte gibt, die keine solchen Kreuzenden Segmente bilden, soll die Methode `false` zurückgeben.
* Testen Sie Ihre Methode, indem Sie die gegebenen 7 Punkte ins korrekte Format bringen, ihrer Methode übergeben, und den Rückgabewert ausgeben.

Für jedes Fenster aus vier aufeinanderfolgenden Punkten wird die vorgegebene Methode `cross` aufgerufen.


````java
class Turns {
    // Diese Methode sollen Sie nicht verändern
    public static boolean cross(double x1, double y1, double x2, double y2, double x3, double y3, double x4, double y4) {
        if (Math.signum(((y2 - y1) * (x3 - x2)) - ((x2 - x1) * (y3 - y2))) != Math.signum(((y2 - y1) * (x4 - x2)) - ((x2 - x1) * (y4 - y2))) &&
            Math.signum(((y4 - y3) * (x2 - x4)) - ((x4 - x3) * (y2 - y4))) != Math.signum(((y4 - y3) * (x1 - x4)) - ((x4 - x3) * (y1 - y4)))) {
            return true;
        }
        return false;
    }
    public static boolean thrackle(double[][] points) {
        for (int i = 0; i + 3 < points.length; i++) {
            double[] p1 = points[i];
            double[] p2 = points[i + 1];
            double[] p3 = points[i + 2];
            double[] p4 = points[i + 3];
            if (!cross(p1[0], p1[1], p2[0], p2[1],
                       p3[0], p3[1], p4[0], p4[1])) {
                return false;
            }
        }
        return true;
    }

    public static void main(String[] args) {
        double[] p1 = {-1.72, -0.18};
        double[] p2 = {1.84, 3.68};
        double[] p3 = {-2.38, 3.98};
        double[] p4 = {1.66, -0.48};
        double[] p5 = {0.08, 5.38};
        double[] p6 = {-0.32, 0.36};
        double[] p7 = {3.44, 2.74};

        double[][] points = {p1, p2, p3, p4, p5, p6, p7};
        System.out.println(thrackle(points));
    }
}

Turns.main(new String[0]);
````

## Aufgabe 10 (5 Punkte)

Schreiben Sie `mode(int[] zahlen)`. Im Array stehen nur Zahlen von `0` bis `9`. Geben Sie die Zahl zurück, die am häufigsten vorkommt. Bei Gleichstand gewinnt die kleinere Zahl. Für ein leeres Array geben Sie `-1` zurück.

Beispiel: Für `{3, 1, 4, 1, 5, 9, 3, 3}` ist das Ergebnis `3`.

````java
class Mode {
    public static int mode(int[] zahlen) {
        if (zahlen.length == 0) {
            return -1;
        }
        int[] haeufigkeit = new int[10];
        for (int zahl : zahlen) {
            haeufigkeit[zahl]++;
        }
        int ergebnis = 0;
        for (int zahl = 1; zahl < haeufigkeit.length; zahl++) {
            if (haeufigkeit[zahl] > haeufigkeit[ergebnis]) {
                ergebnis = zahl;
            }
        }
        return ergebnis;
    }

    public static void main(String[] args) {
        System.out.println(mode(new int[]{3, 1, 4, 1, 5, 9, 3, 3}));
        System.out.println(mode(new int[]{7, 2, 7, 2}));
        System.out.println(mode(new int[]{}));
    }
}

Mode.main(new String[0]);
````
