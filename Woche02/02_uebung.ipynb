## Eine erfundene Sprache

Lesen Sie die zwei Programme und ihre Ausgaben. `text`, `zahl` und `komma` sind Datentypen, `<-` weist zu, `&` verbindet Texte und `%%` beginnt einen Kommentar. `/` liefert ein Kommaergebnis, `//` teilt ganzzahlig. Für Aufgabe 10 liefert `%` den Rest einer Ganzzahldivision.

**Programm 1**
```
Start:
    text wort1 <- 'Hello'
    text wort2 <- 'Dolly'
    text kombi1 <- wort1 & wort2
    text kombi2 <- wort1 & ' ' & wort2
    output(kombi1)
    output(kombi2)
:End
```
Ausgabe: `HelloDolly`, `Hello Dolly`

**Programm 2**
```
Start:
    %% Das ist ein Kommentar
    zahl a <- 5
    zahl b <- 4
    zahl c <- a + b
    zahl d <- a * b
    komma e <- a / b
    komma f <- a // b
    zahl g <- a // b
    output(c, d, e, f, g)
:End
```
Ausgabe: `9, 20, 1.25, 1.0, 1`
