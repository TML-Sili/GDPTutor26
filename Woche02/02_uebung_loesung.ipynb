## Aufgabe 3 – Ausdruck oder Anweisung? (Java)

Markieren Sie im Programm per Kommentar eine vollständige Anweisung und zwei Ausdrücke. Geben Sie zusätzlich `punkte * 2` aus. Welchen Wert hat `punkte` bei der ersten Ausgabe? Warum ist `punkte * 2` allein keine Anweisung?

````java
class Aufgabe03 {
    public static void main(String[] args) {
        int punkte = 4;
        // Die Vorlage 'punkte + 3;' ist in Java keine gültige Anweisung.
        // Unverändert lässt sich die Zelle nicht übersetzen: Es gibt keine erste Ausgabe.
        punkte = punkte + 3; // Gültige Zuweisung; Ausdruck 1 ist punkte + 3.
        System.out.println("Punkte: " + punkte); // Nach der Korrektur: "Punkte: 7".
        // Ausdruck 2: punkte * 2 berechnet 7 * 2, also 14.
        System.out.println("Doppelte Punkte: " + (punkte * 2));
        // punkte * 2 allein berechnet nur einen Wert. Erst eine Ausgabe oder
        // Zuweisung verwendet ihn; der nackte Ausdruck ist keine Java-Anweisung.
    }
}
Aufgabe03.main(new String[0]);
````

## Aufgabe 4 – Sticker verteilen (Java)

`17` Sticker werden auf `5` Personen verteilt. Berechnen und geben Sie aus: (a) ganze Sticker pro Person, (b) übrige Sticker und (c) den rechnerischen Durchschnitt.

````java
class Aufgabe04 {
    public static void main(String[] args) {
        int sticker = 17;
        int personen = 5;
        int ganzeProPerson = sticker / personen;
        int rest = sticker % personen;
        double durchschnitt = (double) sticker / personen;
        System.out.println("Ganze Sticker pro Person: " + ganzeProPerson);
        System.out.println("Übrige Sticker: " + rest);
        System.out.println("Durchschnitt: " + durchschnitt);
    }
}
Aufgabe04.main(new String[0]);
````

## Aufgabe 6 – Ein kleines Programm planen (Java)

Für einen Ausflug stehen `30.0` Franken bereit. `7` Teilnehmende brauchen je ein Ticket zu `3.75` Franken. Schreiben Sie ein Programm mit sinnvoll benannten Variablen. Berechnen Sie Gesamtkosten und verbleibendes Geld. Geben Sie einen verständlichen Satz mit Anlass, Anzahl, Gesamtkosten und Rest aus. Notieren Sie in Kommentaren, welche Zeilen Deklarationen, Zuweisungen und Ausdrücke enthalten. Ändern Sie die Anzahl auf `8`: Was zeigt der Rest dann an, und warum? Benutzen Sie noch keine Bedingungen.

````java
class Aufgabe06 {
    public static void main(String[] args) {
        String anlass = "Ausflug"; // Deklaration mit Zuweisung
        double budget = 30.0;
        int teilnehmende = 7;
        double ticketpreis = 3.75;
        double gesamtkosten = teilnehmende * ticketpreis; // Ausdruck rechts von =
        double rest = budget - gesamtkosten; // Ausdruck rechts von =
        System.out.println(anlass + ": " + teilnehmende + " Tickets kosten "
                + gesamtkosten + " Franken; übrig sind " + rest + " Franken.");
        // Auch die Verkettungen innerhalb von println sind Ausdrücke.
        teilnehmende = 8; // Neue Zuweisung für die zweite Frage.
        gesamtkosten = teilnehmende * ticketpreis;
        rest = budget - gesamtkosten;
        System.out.println("Bei 8 Teilnehmenden bleiben " + rest + " Franken.");
        // 8 * 3.75 = 30.0, also wird das ganze Budget gebraucht.
    }
}
Aufgabe06.main(new String[0]);
````

## Aufgabe 7 – Zwei Werte als Balken darstellen (Turtle)

Stellen Sie `wertA = 18.0` und `wertB = 32.0` als **zwei senkrechte Balken** dar. Mit `faktor = 2.0` soll die Länge jedes Balkens `wert * faktor` sein. Starten Sie beide Balken auf gleicher Höhe, aber mit 30 Schritten Abstand. Zeichnen Sie A blau und B rot (`penColor(BLUE)`/`penColor(RED)`), jeweils mit `penSize(5)`. Zwischen den Balken den Stift mit `penUp()` anheben. Geben Sie zusätzlich einen Satz mit der Differenz `wertB - wertA` aus. Überlegen Sie vor dem Zeichnen, warum beide Balken im 200 × 200 grossen Zeichenbereich sichtbar bleiben. Benutzen Sie nur Variablen, Ausdrücke und die bekannten Turtle-Befehle, keine Schleifen.

````java
class Aufgabe07 {
    public static void main(String[] args) {
        double wertA = 18.0;
        double wertB = 32.0;
        double faktor = 2.0;
        clear();
        reset();
        home();
        penSize(5);
        penColor(BLUE);
        penDown();
        forward(wertA * faktor); // 36 Schritte nach oben
        penUp();
        backward(wertA * faktor); // Zurück zur gemeinsamen Grundlinie
        turnRight(90);
        forward(30);
        turnLeft(90);
        penColor(RED);
        penDown();
        forward(wertB * faktor); // 64 Schritte nach oben
        System.out.println("Differenz: " + (wertB - wertA)); // 14.0
        // Von der Mitte reichen 100 Schritte bis zum Rand; 64 < 100.
        display(drawing());
    }
}
Aufgabe07.main(new String[0]);
````
